```python
# Definición de la clase padre
class Empleado:
    def __init__(self, nombre, salario_base):
        self.nombre = nombre
        self.salario_base = salario_base

    def calcular_pago(self):
        return self.salario_base

    def mostrar_detalles(self):
        print(f"Empleado: {self.nombre}")
        print(f"Puesto: {self.__class__.__name__}")
        print(f"Pago de este mes: ${self.calcular_pago():,.2f}")
        print("-" * 30)

# Creación de clases hija
class Desarrollador(Empleado):
    def __init__(self, nombre, salario_base, lenguaje_principal):
        # Llamada al constructor de la clase base
        super().__init__(nombre, salario_base)
        self.lenguaje_principal = lenguaje_principal

    # Polimorfismo: Redefinición del método calcular_pago
    def calcular_pago(self):
        bono_desarrollador = 2000
        return self.salario_base + bono_desarrollador


class Gerente(Empleado):
    def __init__(self, nombre, salario_base, proyecto_asignado):
        # Llamada al constructor de la clase base
        super().__init__(nombre, salario_base)
        self.proyecto_asignado = proyecto_asignado

    # Polimorfismo: Redefinición del método calcular_pago
    def calcular_pago(self):
        bono_gerente = self.salario_base * 0.15
        return self.salario_base + bono_gerente


if __name__ == "__main__":
    print("--- SISTEMA DE GESTIÓN DE EMPLEADOS ---\n")

    # Creación de instancias (objetos)
    dev1 = Desarrollador("Tomás de Jesús", 15000, "Python")
    gerente1 = Gerente("Paola Rodríguez", 25000, "Plataforma Educativa Immersiva")

    # Lista de empleados (Demostración de Polimorfismo)
    lista_empleados = [dev1, gerente1]

    # Ejecución del ciclo
    for empleado in lista_empleados:
        # Cada objeto ejecuta su propia versión de calcular_pago()
        # a través del método mostrar_detalles()
        empleado.mostrar_detalles()
```

```python
# ADA 4.

# CÓDIGO A REFACTORIZAR (No pueden usar IFs en la solución final)
class GestorEnvios:
    def procesar_envio(self, tipo_transporte, distancia, peso):
        if tipo_transporte == "Drone":
            if distancia > 15:
                return "Error: Excede rango"
            return (distancia * 5) + (peso * 2)
        elif tipo_transporte == "Motocicleta":
            return (distancia * 2) + 10 # Tarifa fija de 10
        elif tipo_transporte == "Camion":
            # Los camiones cobran un 15% extra de impuestos sobre el peso
            return (distancia * 1.5) + (peso * 4 * 1.15)
        else:
            return "Transporte no soportado"
```
