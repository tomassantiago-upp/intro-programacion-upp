```python
class Parrot:
    species = 'bird'

    def __init__(self, name, age, gender, color = 'green'):
        self.name = name
        self.age = age
        self.gender = gender
        self.color = color

    def dance(self):
        print(f"{self.name} está bailando.")

    def greetings(self):
        print(f"{self.name} dice hola.")

blu = Parrot('Blu', 5, "macho")
blu.dance()
blu.greetings()

```
